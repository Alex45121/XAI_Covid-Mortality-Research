```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split, cross_val_predict
from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score
from sklearn.preprocessing import LabelEncoder, StandardScaler

import shap
from imblearn.over_sampling import SMOTE

np.random.seed(42)
```

## Dataset
We'll use the **UCI Adult (Census Income)** dataset from `fetch_openml`. This is a binary classification problem: predict whether a person's income is `>50K` (positive class) or `<=50K`. It is a classic dataset for data-centric work because it contains:
* Real class imbalance (~24% positive)
* A sensitive attribute (`sex`) that we will use for bias analysis
* Categorical features that require encoding
* Natural noise and ambiguity in the labels

> *Note: `fetch_openml` will download the dataset (~4 MB) the first time the cell runs and cache it locally afterwards.*


```python
# Load the dataset
adult = fetch_openml("adult", version=2, as_frame=True)
df = adult.frame.copy()

# Basic inspection
print("Shape:", df.shape)
df.head()
```

### Preprocessing
We drop rows with missing values for simplicity, label-encode categorical variables, and separate features from the target. **Keep the mapping of the `sex` column** — we will need it later for bias analysis.


```python
# Drop rows with missing values
df = df.dropna().reset_index(drop=True)

# Target: >50K  ->  1,  <=50K  ->  0
y = (df["class"] == ">50K").astype(int).values

# Feature matrix (drop target)
X_df = df.drop(columns=["class"])

# Remember the sex column in its original (string) form — we will use it as a sensitive attribute
sex_raw = X_df["sex"].astype(str).values   #  'Male' / 'Female'

# Label-encode all categorical columns
cat_cols = X_df.select_dtypes(include=["object", "category"]).columns.tolist()
encoders = {}
for c in cat_cols:
    le = LabelEncoder()
    X_df[c] = le.fit_transform(X_df[c].astype(str))
    encoders[c] = le

feature_names = X_df.columns.tolist()
X = X_df.values.astype(float)
print("Features:", feature_names)
print("X shape:", X.shape, " y shape:", y.shape)
print("Positive class rate:", y.mean().round(3))
```

Split data into train and test. Fix the random state and use 20% for testing. We also carry the `sex_raw` array through the split so we can group test predictions by gender later.


```python
X_train, X_test, y_train, y_test, sex_train, sex_test = train_test_split(
    X, y, sex_raw, test_size=0.2, random_state=42, stratify=y
)
print("Train:", X_train.shape, "Test:", X_test.shape)
```

### Train a baseline model
We will use a **Random Forest** classifier as our baseline, matching previous labs. All exercises compare against this baseline.


```python
clf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
clf.fit(X_train, y_train)

y_pred = clf.predict(X_test)
print(classification_report(y_test, y_pred, target_names=["<=50K", ">50K"]))
```

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="g", cmap="Blues",
            xticklabels=["<=50K", ">50K"], yticklabels=["<=50K", ">50K"])
plt.xlabel("Predicted"); plt.ylabel("Actual"); plt.title("Baseline confusion matrix")
plt.show()
```

## Exercise 5: Comparing Data Complexity Across Sensitive Groups

In Exercise 3 we observed that the model produces different outcomes for Male and Female groups (DIR well below the 0.8 threshold) and has different per-group accuracies. There are (at least) two possible root causes:

1. **Label imbalance** — the positive rate for the two groups differs in the training data, so the model inherits that disparity.
2. **Intrinsic classification difficulty** — even after controlling for label distribution, the classes may be *geometrically harder to separate* for one group than the other.

Data complexity measures (lecture slide 43) let us quantify root cause #2 entirely from the data, without any model. In this exercise you implement one such measure, compare it across groups, and use the result to explain the bias observed in Ex 3.

**(a)** Implement the **N2 non-parametric separability** measure from lecture slide 43:

$$
\mathrm{N2} = \frac{\sum_{x} d_\text{intra}(x)}{\sum_{x} d_\text{inter}(x)}
$$

where $d_\text{intra}(x)$ is the distance from $x$ to its nearest same-class neighbour, and $d_\text{inter}(x)$ is the distance from $x$ to its nearest different-class neighbour. Smaller N2 means the classes are more locally separable (easier classification problem).

Also implement a second complexity measure of your choice — a natural one is **Fisher's discriminant ratio (F1)** per feature:

$$
F_j = \frac{(\mu_{j,1} - \mu_{j,0})^2}{\sigma_{j,0}^2 + \sigma_{j,1}^2}
$$

and take the maximum over features. High $F$ = at least one feature separates the classes well.

Make sure you **standardise the features** (e.g. `StandardScaler` fit on the *full* training set) before computing distance-based measures, otherwise features with large numeric ranges will dominate the distance.
